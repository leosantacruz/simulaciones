```python
import numpy as np
import matplotlib.pyplot as plt
import random
import time
from IPython.display import clear_output, display

MOVIMIENTOS = [
    [0, 1],
    [0, -1],
    [1, 0],
    [-1, 0]
]


grid = np.zeros((5, 5))
grid[2, 1:4] = -1.0
grid[4, 4] = 1.0
agente = [0,0]
for paso in range(40):
    
    dx, dy = random.choice(MOVIMIENTOS)
    nueva_fila = agente[0] + dx
    nueva_columna = agente[1] + dy
    
    if 0 <= nueva_fila < 5 and 0 <= nueva_columna < 5:
        grid[tuple(agente)] = 0.2
        agente = [nueva_fila,nueva_columna]
        grid[tuple(agente)] = 0.5

        if agente == [4,4]:
            print("Llegó a la meta!!")
            break
        clear_output(wait=True)    
        plt.imshow(grid)
        plt.show()
        time.sleep(0.1)
        
```
